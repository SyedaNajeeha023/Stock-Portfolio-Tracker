# Stock Portfolio Tracker

Use this notebook to track stock investments, current portfolio value, and profit/loss.


```python
import numpy as np
import pandas as pd

try:
    import yfinance as yf
except Exception:
    yf = None
```

```python
def fetch_current_price(ticker, fallback_prices=None):
    fallback_prices = fallback_prices or {}

    if ticker in fallback_prices:
        return float(fallback_prices[ticker])

    if yf is None:
        return np.nan

    try:
        info = yf.Ticker(ticker).history(period='1d')
        if info.empty:
            return np.nan
        return float(info['Close'].iloc[-1])
    except Exception:
        return np.nan
```

```python
# Add as many stocks as needed
portfolio_data = [
    {'Ticker': 'AAPL', 'Quantity': 10, 'Purchase Price': 150.0},
    {'Ticker': 'MSFT', 'Quantity': 5, 'Purchase Price': 300.0},
    {'Ticker': 'GOOGL', 'Quantity': 3, 'Purchase Price': 120.0},
]

# Optional fallback values if yfinance is unavailable
manual_current_prices = {
    'AAPL': 190.0,
    'MSFT': 420.0,
    'GOOGL': 175.0,
}
```

```python
portfolio_df = pd.DataFrame(portfolio_data)
portfolio_df['Invested Value'] = portfolio_df['Quantity'] * portfolio_df['Purchase Price']
portfolio_df['Current Price'] = portfolio_df['Ticker'].apply(lambda t: fetch_current_price(t, manual_current_prices))
portfolio_df['Current Value'] = portfolio_df['Quantity'] * portfolio_df['Current Price']
portfolio_df['Profit/Loss'] = portfolio_df['Current Value'] - portfolio_df['Invested Value']
portfolio_df['Profit/Loss %'] = np.where(
    portfolio_df['Invested Value'] > 0,
    (portfolio_df['Profit/Loss'] / portfolio_df['Invested Value']) * 100,
    np.nan,
)
```

```python
total_invested = portfolio_df['Invested Value'].sum()
current_portfolio_value = portfolio_df['Current Value'].sum()
total_profit_loss = current_portfolio_value - total_invested
profit_loss_percentage = (total_profit_loss / total_invested) * 100 if total_invested else np.nan
```

```python
display(portfolio_df)
print(f'Total Investment Value: {total_invested:.2f}')
print(f'Current Portfolio Value: {current_portfolio_value:.2f}')
print(f'Profit / Loss: {total_profit_loss:.2f}')
print(f'Profit / Loss Percentage: {profit_loss_percentage:.2f}%')
```
