```python
# --- Stock Portfolio Tracker ---

# Hardcoded stock prices dictionary
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "AMZN": 185,
    "MSFT": 420
}

print("Available stocks:", list(stock_prices.keys()))
print("-" * 35)
# --- Input: stock name and quantity ---

portfolio = {}

while True:
    stock = input("Enter stock name (or 'done' to finish): ").upper()
    if stock == "DONE":
        break
    if stock not in stock_prices:
        print(f"'{stock}' not found. Available: {list(stock_prices.keys())}")
        continue
    qty = int(input(f"Enter quantity for {stock}: "))
    portfolio[stock] = portfolio.get(stock, 0) + qty

print("\nPortfolio entered:", portfolio)
# --- Calculate total investment ---

print("\n--- Portfolio Summary ---")
print(f"{'Stock':<8} {'Qty':>5} {'Price':>8} {'Value':>10}")
print("-" * 35)

total = 0
for stock, qty in portfolio.items():
    price = stock_prices[stock]
    value = price * qty
    total += value
    print(f"{stock:<8} {qty:>5} {price:>8} {value:>10}")

print("-" * 35)
print(f"{'TOTAL INVESTMENT':>29} ${total:>9,.2f}")
# --- Optional: Save to file ---

save = input("\nSave results? Enter 'txt', 'csv', or 'no': ").lower()

if save == "txt":
    with open("portfolio.txt", "w") as f:
        f.write("Stock Portfolio Summary\n")
        f.write("-" * 35 + "\n")
        for stock, qty in portfolio.items():
            value = stock_prices[stock] * qty
            f.write(f"{stock}: {qty} shares = ${value}\n")
        f.write(f"Total: ${total:.2f}\n")
    print("Saved to portfolio.txt")

elif save == "csv":
    import csv
    with open("portfolio.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["Stock", "Quantity", "Price", "Value"])
        for stock, qty in portfolio.items():
            price = stock_prices[stock]
            writer.writerow([stock, qty, price, price * qty])
        writer.writerow(["", "", "Total", total])
    print("Saved to portfolio.csv")

else:
    print("Not saved.")
```
